```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# =========================
# 1. Preparação dos dados
# =========================
data = load_iris()
X = data.data
y = data.target

scaler = StandardScaler()
X = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

X_train = torch.tensor(X_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)

# =========================
# 2. Modelo MLP
# =========================
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(4, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8, 3)
        )

    def forward(self, x):
        return self.model(x)

model = MLP()

# =========================
# 3. Treinamento
# =========================
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

epochs = 100

losses = []
accuracies = []

for epoch in range(epochs):
    model.train()

    outputs = model(X_train)
    loss = criterion(outputs, y_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    losses.append(loss.item())

    # Acurácia
    _, predicted = torch.max(outputs, 1)
    acc = (predicted == y_train).float().mean()
    accuracies.append(acc.item())

    print(f"Epoch {epoch+1}: Loss={loss.item():.4f} | Acc={acc.item():.4f}")

# =========================
# 4. Avaliação no teste
# =========================
model.eval()
with torch.no_grad():
    outputs = model(X_test)
    _, predicted = torch.max(outputs, 1)
    test_acc = (predicted == y_test).float().mean()

print(f"\nAcurácia no teste: {test_acc:.4f}")

# =========================
# 5. Gráficos (SALVANDO)
# =========================

# Loss
plt.figure()
plt.plot(losses)
plt.title("Loss por época")
plt.xlabel("Épocas")
plt.ylabel("Loss")
plt.savefig("loss.png")
plt.close()

# Acurácia
plt.figure()
plt.plot(accuracies)
plt.title("Acurácia por época")
plt.xlabel("Épocas")
plt.ylabel("Acurácia")
plt.savefig("accuracy.png")
plt.close()

print("\nGráficos salvos como 'loss.png' e 'accuracy.png'")
```
